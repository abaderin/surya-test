```python
from surya.recognition import RecognitionPredictor
from surya.foundation import FoundationPredictor
from surya.detection import DetectionPredictor
from surya.settings import settings

fp = FoundationPredictor(checkpoint=settings.RECOGNITION_MODEL_CHECKPOINT)
dp = DetectionPredictor(checkpoint=settings.DETECTOR_MODEL_CHECKPOINT)
rp = RecognitionPredictor(fp)
```

```python
pdf_path = 'Screenshot 2026-05-22 at 04.36.47.png'
from pathlib import Path
Path(pdf_path).exists()
```

```python
import fitz
from PIL import Image
import io

img = Image.open(pdf_path)

img
```

```python
recognition_result = rp([img], det_predictor=dp, math_mode=True)
```
